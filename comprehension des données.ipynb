## Explication des variables :

- smoking =	Statut tabagique
- diabetes =	Indicateur de diabète
- age =	Âge en années
- age_normalized =	Âge normalisé
- bmi =	Indice de masse corporelle
- hypertension =	Indicateur d'hypertension
- heart_disease =	Indicateur de maladie cardiaque
- HbA1c_level =	Taux d'HbA1c
- glucose =	Taux de glucose sanguin
- cholesterol =	Taux de cholestérol total
- sleep_hours =	Durée quotidienne de sommeil
- triglycerides =	Taux de triglycérides
- physical_activity =	Niveau d'activité physique
- family_history =	Antécédents familiaux
- stress_level =	Niveau de stress
- low_hdl_cholesterol =	Indicateur de faible HDL
- high_ldl_cholesterol =	Indicateur de LDL élevé
- blood_pressure =	Pression artérielle
- high_blood_pressure =	Indicateur de pression artérielle élevée
- sugar_consumption =	Consommation de sucre
- crp_level =	Taux de CRP
- homocysteine_level =	Taux d'homocystéine
- systolic_bp =	Pression artérielle systolique
- diastolic_bp = 	Pression artérielle diastolique
- alcohol_intake =	Consommation d'alcool
- salt_intake =	Consommation de sel
- heart_rate =	Fréquence cardiaque
- hdl =	Taux de HDL
- ldl =	Taux de LDL
- education_level =	Niveau d'éducation
- employment_status =	Statut professionnel

## Classe de sublabel:
- N         -> Aucun indicateur de maladie
- DI        -> Diabète uniquement
- HY        -> Hypertension uniquement
- HT        -> Maladie cardiaque uniquement
- DI_HY     -> Diabète + Hypertension
- DI_HT     -> Diabète + Maladie cardiaque
- HT_HY     -> Maladie cardiaque + Hypertension
- DI_HT_HY  -> Les trois affections combinées
